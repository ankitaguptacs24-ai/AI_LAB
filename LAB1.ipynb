### Vacuum Cleaner Simulation

This simulation models a vacuum cleaner operating in multiple rooms. Each room has a status (dirty or clean). The vacuum cleaner cleans dirty rooms and moves to another room.

#### **Demonstration:**

Let's create some rooms, some clean and some dirty, and see the vacuum cleaner in action.

```python
num_rooms_input = input("Enter the number of rooms to simulate (e.g., 5): ")
try:
    num_rooms = int(num_rooms_input)
    if num_rooms <= 0:
        raise ValueError("Number of rooms must be positive.")
except ValueError as e:
    print(f"Invalid input: {e}. Defaulting to 3 rooms.")
    num_rooms = 3

user_rooms = []
for i in range(num_rooms):
    room_name = chr(ord('A') + i) # Assign A, B, C... names
    while True:
        status_input = input(f"Is Room {room_name} dirty or clean? (d/c): ").lower()
        if status_input == 'd':
            is_dirty = True
            break
        elif status_input == 'c':
            is_dirty = False
            break
        else:
            print("Invalid input. Please enter 'd' for dirty or 'c' for clean.")

    user_rooms.append(Room(room_name, is_dirty))

print("\nInitial room statuses (from your input):")
for room in user_rooms:
    print(room)

user_vacuum_bot = VacuumCleaner(user_rooms)
user_vacuum_bot.operate(cycles=num_rooms * 2) # Operate for twice the number of rooms

print("\nFinal room statuses (from your input):")
for room in user_rooms:
    print(room)
```

### Tic-Tac-Toe Game

This section implements a simple two-player Tic-Tac-Toe game.

```python
def create_board():
    return [' ' for _ in range(9)]  # 9 cells for a 3x3 board

def display_board(board):
    print("\n")
    print(f"{board[0]} | {board[1]} | {board[2]}")
    print("----------")
    print(f"{board[3]} | {board[4]} | {board[5]}")
    print("----------")
    print(f"{board[6]} | {board[7]} | {board[8]}")
    print("\n")

def check_win(board, player):
    # Check rows
    for i in range(0, 9, 3):
        if board[i] == board[i+1] == board[i+2] == player:
            return True
    # Check columns
    for i in range(3):
        if board[i] == board[i+3] == board[i+6] == player:
            return True
    # Check diagonals
    if board[0] == board[4] == board[8] == player:
        return True
    if board[2] == board[4] == board[6] == player:
        return True
    return False

def check_draw(board):
    return ' ' not in board

def get_player_move(player, board):
    while True:
        try:
            move = int(input(f"Player {player}, enter your move (1-9): ")) - 1
            if 0 <= move <= 8 and board[move] == ' ':
                return move
            else:
                print("Invalid move. That position is already taken or out of range. Try again.")
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 9.")
```

```python
def play_game():
    board = create_board()
    current_player = 'X'
    game_over = False

    while not game_over:
        display_board(board)
        move = get_player_move(current_player, board)
        board[move] = current_player

        if check_win(board, current_player):
            display_board(board)
            print(f"Player {current_player} wins!")
            game_over = True
        elif check_draw(board):
            display_board(board)
            print("It's a draw!")
            game_over = True
        else:
            current_player = 'O' if current_player == 'X' else 'X'

# Start the game
play_game()
```
