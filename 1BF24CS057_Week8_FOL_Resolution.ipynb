### Case Study 2: John Likes Peanuts

**Premises:**
1. John likes all kind of food: $\sim Food(x) \lor Likes(John, x)$
2. Apple and vegetables are food: $Food(Apple)$, $Food(Vegetables)$
3. Anything anyone eats and not killed is food: $\sim Eats(y, z) \lor Killed(y) \lor Food(z)$
4. Anil eats peanuts and still alive: $Eats(Anil, Peanuts)$, $Alive(Anil)$
5. Harry eats everything that Anil eats: $\sim Eats(Anil, w) \lor Eats(Harry, w)$
6. Anyone who is alive implies not killed: $\sim Alive(u) \lor ~Killed(u)$

**Query:**
* John likes peanuts: $Likes(John, Peanuts)$

```python
def parse(s):
    result=set()
    for item in s.split("|"):
        item=item.strip()
        neg=item.startswith("~")
        if neg:
            item=item[1:].strip()
        p,args=item.split("(",1)
        args=tuple(x.strip() for x in args.rstrip(")").split(","))
        result.add((neg,p.strip(),args))
    return frozenset(result)

def isvar(x):
    return x.startswith("?")

def unify(x,y,sub):
    if x in sub:
        x=sub[x]
    if y in sub:
        y=sub[y]
    if x==y:
        return sub
    if isvar(x):
        sub[x]=y
        return sub
    if isvar(y):
        sub[y]=x
        return sub
    return None

def replace(c,sub):
    return frozenset(
        (n,p,tuple(sub.get(x,x) for x in a))
        for n,p,a in c
    )

def resolve(c1,c2):
    for l1 in c1:
        for l2 in c2:
            if l1[0]!=l2[0] and l1[1]==l2[1]:
                if len(l1[2])!=len(l2[2]):
                    continue
                sub={}
                for x,y in zip(l1[2],l2[2]):
                    sub=unify(x,y,sub)
                    if sub is None:
                        break
                if sub is not None:
                    return replace((c1-{l1})|(c2-{l2}),sub)
    return None

def resolution(kb):
    clauses=set(kb)
    for step in range(1000):
        old=list(clauses)
        new=set()
        for i in range(len(old)):
            for j in range(i+1,len(old)):
                r=resolve(old[i],old[j])
                if r is not None:
                    if not r:
                        print("Empty clause obtained!")
                        print("Query is PROVED")
                        return
                    if r not in clauses:
                        new.add(r)
        if not new:
            break
        clauses.update(new)
    print("Query could not be proved")

# Taking input from user
kb=[]
n=int(input("Enter number of KB clauses: "))

for i in range(n):
    c=input("Enter clause "+str(i+1)+": ")
    kb.append(parse(c))

q=input("Enter query: ").strip()

# Negating the query
if q.startswith("~"):
    q=q[1:]
else:
    q="~"+q

kb.append(parse(q))

print("\nApplying Resolution...")
resolution(kb)
```
