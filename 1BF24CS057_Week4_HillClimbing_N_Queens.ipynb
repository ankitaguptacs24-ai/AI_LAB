```python
import random

def print_board(state):
    """Prints a compact text-based representation of the board."""
    n = len(state)
    for row in range(n):
        line = ["Q" if state[col] == row else "." for col in range(n)]
        print(" ".join(line))
    print()

def get_attacks(state):
    """Calculates attacking pairs of queens."""
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def solve_n_queens_hill_climbing(n=4):
    print(f"--- Solving {n}-Queens using Hill Climbing ---\n")
    step = 0
    while True:
        # Generate a random initial state
        current_state = tuple(random.randint(0, n - 1) for _ in range(n))
        current_cost = get_attacks(current_state)

        path = [(current_state, current_cost)]

        # Local search
        while True:
            neighbors = []
            for col in range(n):
                for row in range(n):
                    if current_state[col] != row:
                        neigh = list(current_state)
                        neigh[col] = row
                        neighbors.append(tuple(neigh))

            neighbor_costs = {neigh: get_attacks(neigh) for neigh in neighbors}
            best_cost = min(neighbor_costs.values())
            best_neighbors = [state for state, cost in neighbor_costs.items() if cost == best_cost]

            if best_cost >= current_cost:
                break

            current_state = random.choice(best_neighbors)
            current_cost = best_cost
            path.append((current_state, current_cost))

        if current_cost == 0:
            # Print steps of successful path compactly
            for idx, (state, cost) in enumerate(path):
                print(f"Step {idx} | State: {state} | Cost (Attacks): {cost}")
                print_board(state)
            break

n=int(input("Enter the number of queens: "))
solve_n_queens_hill_climbing(n)
```
